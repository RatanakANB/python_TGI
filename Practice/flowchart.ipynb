````python
def f(x):
    fx = (2 * x) - 2
    if fx == 0:
        return f"x = {x} is a zero of the function (f(x) = {fx})"
    else:
        return f"x = {x} is not a zero of the function (f(x) = {fx})"

x = float(input("Enter a value for x: "))
print(f(x))
````

```mermaid
flowchart LR
    Start([Start]) --> Input[Get user input: x]
    Input --> CallFunc["Call f(1)"]
    CallFunc --> Compute["Calculate fx = 2*x - 2"]
    Compute --> Decision{"fx > 0 ?"}
    
    Decision -- Yes --> NotZero["Return: 'x = fx is not a zero of the function'"]
    Decision -- No --> IsZero["Return: 'x = fx is a zero of the function'"]
    
    NotZero --> Print[Print Result]
    IsZero --> Print
    Print --> End([End])
```
